## Population model — Meta-analytic (Lin & Yao, 2022)

The primary population model uses factor loadings derived from the meta-analysis by Lin and Yao (2022). Four factors: Psychological (6 items), Physical (7 items), Social (3 items), Environment (8 items). Interfactor correlations are set at the upper bound (0.3) reported in the meta-analysis (range: 0.08–0.30). Residual variances follow the formula 1 − λ², except for Q3 and Q4 which share a correlated residual (0.3).

```
#| label: pop-model

cat(popModel)
```

## Naive model — Minimally acceptable scenario

All factor loadings set to 0.5 (minimum acceptable for CFA), residual variances = 0.75 (i.e., 1 − 0.5²), and small interfactor correlations (0.3). This represents a **pessimistic** scenario.

```
#| label: naive-model

cat(naiveModel)
```

## Optimistic model — Best-case scenario

All factor loadings set to 0.7 (strong loadings), residual variances = 0.51 (1 − 0.7²), and moderate interfactor correlations (0.5). This represents an **optimistic** scenario.

```
#| label: opt-model

cat(optModel)
```

## Analysis model — Free parameters

This is the model the researcher fits to data. All parameters are freely estimated (no fixed values), matching the structure of the population models.

```
#| label: analysis-model

cat(analyzeModel)
```
