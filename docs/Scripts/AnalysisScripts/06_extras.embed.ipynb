## Dynamic Fit Index (DFI) — Brief Example

Traditional fit index cutoffs (e.g., RMSEA \< .06, CFI \> .95) are generic rules of thumb that do not account for model-specific characteristics such as the number of factors and indicators, the magnitude of loadings, or the degree of model misspecification. The Dynamic Fit Index (DFI) approach \[@mcneish2021\] generates model-specific cutoffs via simulation.

The `dynamic` package implements this method. Below is a brief demonstration using a simplified version of our WHOQOL-BREF model.

```
#| label: dfi-example
#| eval: false

# Install dynamic if needed:
# install.packages("dynamic")
library(dynamic)

# Simple DFI example using a 2-factor subset (for speed)
# In practice, you would use the full model
dfi_model <- '
  psycho =~ Q5_P + Q6_P + Q7_P + Q11_P + Q19_P + Q26_P
  physical =~ Q3_F + Q4_F + Q10_F + Q15_F + Q16_F + Q17_F + Q18_F
'

# cfaHB generates model-specific cutoffs for continuous indicators
# n = target sample size; plot = TRUE for visualization
dfi_result <- dynamic::cfaHB(
  model  = dfi_model,
  n      = 500,
  plot   = TRUE,
  manual = TRUE
)

# Display DFI-derived cutoffs
print(dfi_result)
```

The DFI cutoffs are **specific to this model and sample size**. They will typically be more stringent than the generic rules of thumb for models with few indicators per factor or weak loadings, and more lenient for models with many strong indicators.

For a complete DFI tutorial with the full WHOQOL-BREF model (including ordinal indicators and the `catHB()` function for categorical data), see the companion manuscript: [CFA Best Practices with the WHOQOL-BREF](../cfa-brm/).
